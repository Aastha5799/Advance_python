## Question 6: List Sanitization and Metrics
Given:
```python
marks = [45, -3, 88, "NA", 102, None, 67]
```

Write a function that:
1. Removes invalid entries  
2. Returns:
   - Cleaned list  
   - Number of invalid entries  
   - Percentage of valid marks  

Return values using a tuple.


````python
marks = [45, -3, 88, "NA", 102, None, 67]
def sanitize_marks(marks_list):
    cleaned = []  # To store valid marks
    invalid_count = 0  # Counter for invalid entries

    for mark in marks_list:
        # Check if mark is an integer and between 0 and 100
        if isinstance(mark, int) and 0 <= mark <= 100:
            cleaned.append(mark)
        else:
            invalid_count += 1

    # Calculate percentage of valid marks
    if len(marks_list) > 0:
        valid_percentage = (len(cleaned) / len(marks_list)) * 100
    else:
        valid_percentage = 0

    # Return values as a tuple
    return (cleaned, invalid_count, valid_percentage)

result = sanitize_marks(marks)
print("Cleaned marks:", result[0])
print("Number of invalid entries:", result[1])
print("Percentage of valid marks:", result[2])
````

## Question 8: Prime Number Function
Write `is_prime(n)` and display prime numbers between 1 and 50.


````python
def is_prime(n):
    if n <= 1:
        return False  # 1 and numbers below are not prime
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False  # divisible by a number other than 1 and itself
    return True  # if no divisors found, it's prime
print("Prime numbers between 1 and 50:")
for num in range(1, 51):
    if is_prime(num):
        print(num, end=" ")
````

## Question 9: Function Composition
Write functions to:
1. Calculate total marks  
2. Calculate average  
3. Classify result (Distinction / Pass / Fail)

Then write a function that combines them and returns:
```
(name, total, average, result)
```

Explain why modular design is beneficial.


````python
marks = [85, 76, 90, 65, 70]  # Example marks
name = "Aastha"
def total_marks(marks_list):
    return sum(marks_list)  # Sum all marks
def average_marks(marks_list):
    return sum(marks_list) / len(marks_list)  # Average = total / number of subjects
def classify_result(avg):
    if avg >= 75:
        return "Distinction"
    elif avg >= 40:
        return "Pass"
    else:
        return "Fail"
def student_report(name, marks_list):
    total = total_marks(marks_list)
    avg = average_marks(marks_list)
    result = classify_result(avg)
    return (name, total, avg, result)
report = student_report(name, marks)
print("Student Report:", report)
````

Why modular design is beneficial?

Readability: Each function does one specific task which means easy to understand.

Reusability: Functions like total_marks() or classify_result() can be reused for other students.

Debugging: If there is a mistake, you only need to check one function.

Scalability: You can easily add more functionality (like grade letters) without rewriting everything.

## Question 10: Defensive Input Parsing
Write a function that accepts a raw input string in the format:

```
"Ram|101|22|78, 88, 91"
```
Hint: Parse student data from string format: "Name|Roll|Age|Mark1, Mark2, Mark3"

Tasks:
1. Extract name, roll number, age, and marks  
2. Convert them into appropriate data types  
3. Validate age (≥16) and marks (0–100)  
4. Return the result as a dictionary using roll number as the key  

The function must not crash for malformed input.


````python
def parse_student_data(raw_input):
    result = {}

    try:
        # Split the input using '|'
        parts = raw_input.split("|")
        if len(parts) != 4:
            return result   # malformed input

        # Extract and clean values
        name = parts[0].strip()
        roll = int(parts[1].strip())
        age = int(parts[2].strip())

        # Validate age
        if age < 16:
            return result

        # Parse marks
        marks_list = parts[3].split(",")
        marks = []

        for m in marks_list:
            try:
                mark = int(m.strip())
                if 0 <= mark <= 100:
                    marks.append(mark)
            except ValueError:
                continue   # skip invalid mark

        if len(marks) == 0:
            return result

        # Store result using roll number as key
        result[roll] = {
            "name": name,
            "age": age,
            "marks": marks
        }

    except Exception:
        return result   # prevents crashing for any unexpected error

    return result


# ---------------- TESTING ----------------
if __name__ == "__main__":
    input_string = "Ram|101|22|78, 88, 91"
    output = parse_student_data(input_string)
    print(output)
````

## Question 11: Type Stability and Conversion
Given the list:
```python
data = [10, "20", 30.5, "abc", None, True]
```

Write a function that:
1. Separates numeric and non-numeric values  
2. Converts numeric values to float  
3. Returns both lists  

Explain in comments why `True` behaves as a numeric value.


````python
def separate_and_convert(data):
    numeric_values = []
    non_numeric_values = []

    for item in data:
        # In Python, bool is a subclass of int:
        # True behaves like 1 and False behaves like 0
        # That is why isinstance(True, int) returns True
        if isinstance(item, bool):
            numeric_values.append(float(item))

        # Check for int or float
        elif isinstance(item, (int, float)):
            numeric_values.append(float(item))

        # Check for numeric strings
        elif isinstance(item, str):
            try:
                numeric_values.append(float(item))
            except ValueError:
                non_numeric_values.append(item)

        # All other types (None, objects, etc.)
        else:
            non_numeric_values.append(item)

    return numeric_values, non_numeric_values


# ---------------- TESTING ----------------
if __name__ == "__main__":
    
    data = [10, "20", 30.5, "abc", None, True]

    nums, non_nums = separate_and_convert(data)

    print("Numeric values:", nums)
    print("Non-numeric values:", non_nums)

# True behaves as numeric because:
# bool is a subclass of int in Python
# True == 1 and False == 0
````

## Question 12: Pattern-Based Looping
Generate the sequence:
```
2, 6, 12, 20, 30, 42, 56
```

Rules:
- Do not hardcode the list  
- Use loops and conditionals only  
- Explain the pattern in comments


````python
def generate_sequence():
    sequence = []

    # Pattern explanation:
    # The sequence follows the formula:
    # n * (n + 1)
    # where n starts from 1
    #
    # 1 × 2 = 2
    # 2 × 3 = 6
    # 3 × 4 = 12
    # 4 × 5 = 20
    # 5 × 6 = 30
    # 6 × 7 = 42
    # 7 × 8 = 56

    for n in range(1, 8):
        value = n * (n + 1)
        sequence.append(value)

    return sequence


# ---------------- TESTING ----------------
if __name__ == "__main__":
    result = generate_sequence()
    print(", ".join(map(str, result)))
````

## Question 13: Set-Based Enrollment Analysis
Given:
```python
python = {"Amit", "Sita", "Hari", "Nita"}
sql = {"Hari", "Ram", "Nita"}
statistics = {"Amit", "Hari", "Gita"}
```

Find:
1. Students enrolled in exactly two subjects  
2. Students enrolled in only one subject  
3. Students enrolled in all subjects


````python
# Given sets
python = {"Amit", "Sita", "Hari", "Nita"}
sql = {"Hari", "Ram", "Nita"}
statistics = {"Amit", "Hari", "Gita"}

# -------------------------------
# 1. Students enrolled in ALL subjects
# Intersection of all three sets
all_subjects = python & sql & statistics

# -------------------------------
# 2. Students enrolled in EXACTLY TWO subjects
# (In at least two) minus (in all three)

python_sql = python & sql
python_stats = python & statistics
sql_stats = sql & statistics

exactly_two = (python_sql | python_stats | sql_stats) - all_subjects

# -------------------------------
# 3. Students enrolled in ONLY ONE subject
# (In any subject) minus (in two or more)

all_students = python | sql | statistics
only_one = all_students - (exactly_two | all_subjects)

# -------------------------------
# Output
print("Students enrolled in exactly two subjects:", exactly_two)
print("Students enrolled in only one subject:", only_one)
print("Students enrolled in all subjects:", all_subjects)
````

## Question 14: Dictionary Aggregation Logic
Given:
```python
records = {
    "Amit": [78, 82, 91],
    "Sita": [88, 79, 85],
    "Hari": [45, 52, 60]
}
```

Write a function that:
1. Computes average per student  
2. Identifies top and bottom performers  
3. Returns results using a single dictionary


````python
def analyze_records(records):
    result = {}

    # Dictionary to store average marks of each student
    averages = {}

    # 1. Compute average per student
    for student, marks in records.items():
        if len(marks) == 0:
            avg = 0
        else:
            avg = sum(marks) / len(marks)
        averages[student] = avg

    # 2. Identify top and bottom performers
    top_student = max(averages, key=averages.get)
    bottom_student = min(averages, key=averages.get)

    # 3. Store everything in a single dictionary
    result["averages"] = averages
    result["top_performer"] = {
        "name": top_student,
        "average": averages[top_student]
    }
    result["bottom_performer"] = {
        "name": bottom_student,
        "average": averages[bottom_student]
    }

    return result


# ---------------- TESTING ----------------
if __name__ == "__main__":
    records = {
        "Amit": [78, 82, 91],
        "Sita": [88, 79, 85],
        "Hari": [45, 52, 60]
    }

    analysis = analyze_records(records)

    print(analysis)
````

## Question 15: Nested Data Structure Analysis
Given a nested dictionary representing a school's grade book:
```python
school_data = {
    "Class A": {
        "Amit": [78, 82, 91],
        "Sita": [88, 79, 85]
    },
    "Class B": {
        "Hari": [45, 52, 60],
        "Ram": [92, 88, 95]
    }
}
```

Write a function that:
1. Calculates the average marks for each student
2. Finds the top performer in each class
3. Calculates the overall school average
4. Returns a dictionary with:
   - `class_toppers`: {class_name: (student_name, average)}
   - `school_average`: float
   - `student_averages`: {student_name: average}

Explain in comments why nested dictionaries are useful for hierarchical data.

````python
def analyze_school_data(school_data):
    # Nested dictionaries are useful for hierarchical data because:
    # They allow data to be grouped logically at multiple levels.
    # Here:
    #   - The outer dictionary represents classes
    #   - The inner dictionary represents students in each class
    # This structure makes it easy to access, update, and analyze
    # data at any level (class-wise or student-wise).

    class_toppers = {}
    student_averages = {}

    total_marks = 0
    total_count = 0

    # Loop through each class
    for class_name, students in school_data.items():
        highest_avg = -1
        topper_name = ""

        # Loop through each student in the class
        for student, marks in students.items():
            # Calculate average marks for student
            avg = sum(marks) / len(marks)

            # Store student average
            student_averages[student] = avg

            # Update total marks and count for school average
            total_marks += sum(marks)
            total_count += len(marks)

            # Check for class topper
            if avg > highest_avg:
                highest_avg = avg
                topper_name = student

        # Store class topper information
        class_toppers[class_name] = (topper_name, highest_avg)

    # Calculate overall school average
    school_average = total_marks / total_count if total_count != 0 else 0

    # Return the aggregated result
    return {
        "class_toppers": class_toppers,
        "school_average": school_average,
        "student_averages": student_averages
    }


# ---------------- TESTING ----------------
if __name__ == "__main__":
    school_data = {
        "Class A": {
            "Amit": [78, 82, 91],
            "Sita": [88, 79, 85]
        },
        "Class B": {
            "Hari": [45, 52, 60],
            "Ram": [92, 88, 95]
        }
    }

    result = analyze_school_data(school_data)
    print(result)
````
